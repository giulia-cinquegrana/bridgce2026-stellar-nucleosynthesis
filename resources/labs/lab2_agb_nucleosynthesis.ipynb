# Diagnosing Neutron-Capture Regimes of AGB Ejecta 

AGBs have rich nucleosynthesis, but the final signature of the AGB material is very sensitive to the initial mass and metallicity of the progenitor. This is what makes them powerful for diagnosing stellar information. As we covered in the previous session, each element usually has some contributions from multiple sources. So the strongest diagnostics have to come at the isotopic level. Here, we are going to focus on distinguishing between the s- and i-processes in AGBs. To do so, the barium isotopes are paticularly valuable.

## Why Ba? 

Barium's five s-process-reachable isotopes ($^{134-138}\mathrm{Ba}$) are produced through a short chain of radioactive caesium isotopes, each of which can act as a branch point. At a branch point, neutron capture competes with $\beta$-decay, making the resulting isotope pattern a thermometer and densitometer of the stellar environment. A branch opens when the local neutron density is high enough for neutron capture to compete with $\beta$-decay on the isotope's half-life. The pattern that emerges therefore records the conditions—and ultimately the stellar mechanism—in which the nuclei were produced. There are three branch points in this chain:

$$
^{133}\mathrm{Cs}\;(n,\gamma)\rightarrow\; {}^{134}\mathrm{Cs}
\;\underset{\text{decay}}{\overset{\text{capture}}{\rightleftarrows}}\;
\begin{cases}
\beta^- \rightarrow {}^{134}\mathrm{Ba}\\[4pt]
(n,\gamma)\rightarrow {}^{135}\mathrm{Cs}
\;\underset{\text{decay}}{\overset{\text{capture}}{\rightleftarrows}}\;
\begin{cases}
\beta^- \rightarrow {}^{135}\mathrm{Ba}\\[4pt]
(n,\gamma)\rightarrow {}^{136}\mathrm{Cs}
\;\underset{\text{decay}}{\overset{\text{capture}}{\rightleftarrows}}\;
\begin{cases}
\beta^- \rightarrow {}^{136}\mathrm{Ba}\\[4pt]
(n,\gamma)\rightarrow {}^{137}\mathrm{Cs} \;\xrightarrow{\beta^-}\; {}^{137}\mathrm{Ba}
\end{cases}
\end{cases}
\end{cases}
$$

The barium isotopes formed can themselves keep capturing neutrons, $^{134}\mathrm{Ba}\to{}^{135}\mathrm{Ba}\to{}^{136}\mathrm{Ba}\to{}^{137}\mathrm{Ba}\to{}^{138}\mathrm{Ba}$, with $^{138}\mathrm{Ba}$ a small-cross-section bottleneck ($N=82$ shell closure). Each of the three branch points has its own $\beta$-decay half-life, so each responds to the ambient neutron density at a different threshold — and it is this *staggering* of thresholds that lets the isotope pattern (not the elemental Ba/Fe or Ba/Eu ratio) fingerprint the regime:

- $^{135}\mathrm{Cs}$ is essentially immortal on stellar timescales ($t_{1/2}\approx2.3\times10^6$ yr). Its branch is already almost fully capture-dominated even at ordinary *s*-process densities, so the route to $^{135}\mathrm{Ba}$ does not change significantly between *s*- and *i*-process conditions.
- $^{134}\mathrm{Cs}$ ($t_{1/2}=2.06$ yr) starts shifting toward capture already at *s*-process-like densities, so $^{134}\mathrm{Ba}$ production is partially suppressed even in the classical *s*-process.
- The decay of $^{136}\mathrm{Cs}$ ($t_{1/2}=13.16$ d) can quite easily outrun capture at *s*-process densities, so essentially all flow through it becomes $^{136}\mathrm{Ba}$ and the chain never reaches $^{137}\mathrm{Cs}$. Typically, only at the much higher densities reached by the *i*-process does capture start to win here too and open a path to $^{137}\mathrm{Cs}\to{}^{137}\mathrm{Ba}$ that is otherwise closed.

This does *not* mean $^{137}\mathrm{Ba}/^{138}\mathrm{Ba}$ is zero at *s*-process densities and switches on at *i*-process densities — the $N=82$ bottleneck at $^{138}\mathrm{Ba}$ (its capture cross section is tiny) already produces some pile-up at $^{137}\mathrm{Ba}$ regardless of density, so the ratio starts from a non-trivial baseline. What actually distinguishes the two regimes is a *crossover*: as $n_n$ rises, $^{134}\mathrm{Ba}/^{138}\mathrm{Ba}$ and $^{135}\mathrm{Ba}/^{138}\mathrm{Ba}$ first rise, then collapse toward a low floor once the $^{134}\mathrm{Cs}$ branch fully closes, while $^{137}\mathrm{Ba}/^{138}\mathrm{Ba}$ keeps climbing as the $^{136}\mathrm{Cs}$ branch opens — so at *i*-process densities $^{137}\mathrm{Ba}/^{138}\mathrm{Ba}$ pulls clearly away from $^{134,135}\mathrm{Ba}/^{138}\mathrm{Ba}$, rather than appearing from nothing. You will compute this directly in Exercises 1–2 below, and see why $^{135}\mathrm{Ba}$, despite also sitting at a branch point, never develops that separation.

## Nuclear-physics tools

The **neutron capture rate** for a target nucleus is given by:

$$
\lambda_{n,i} = n_n \langle \sigma v\rangle_i = n_n\, \sigma_{30,i}\, v_T
$$

where: 
- $n_n$ is the neutron density
- $kT$ is the thermal energy
- $\sigma_{30,i}$ is the Maxwellian-averaged cross section at $kT=30$ keV
- $v_T=\sqrt{2kT/m_n}=2.194\times10^{8}\ \mathrm{cm\,s^{-1}}$ is the corresponding thermal velocity. 

From the terrestrial half-life, the **$\beta$ decay rate** is given by $\lambda_{\beta,i}=\ln2/t_{1/2,i}$ (we will ignore the temperature-dependence of stellar β-decay rates here). At a branch-point nucleus then, the fraction of flow that decays before capturing another neutron is:

$$
f_\beta = \frac{\lambda_\beta}{\lambda_\beta + \lambda_n}\ ,\qquad f_n = \frac{\lambda_n}{\lambda_\beta+\lambda_n} = 1-f_\beta .
$$

Since $\lambda_n \propto n_n$ and $\lambda_\beta$ is fixed, raising $n_n$ pushes the branching ratio toward the neutron-capture channel. Different branch points have different $\lambda_\beta$, so they respond to a given $n_n$ differently.

To compare a low-density, long-duration process (classical *s*-process, $n_n\sim10^{7\text{–}9}\,\mathrm{cm^{-3}}$ sustained for $\sim10^{3\text{-}4}$ yr) with a high-density, short-duration process (proposed *i*-process, $n_n\gtrsim10^{13\text{–}17}\,\mathrm{cm^{-3}}$ sustained for hours–days) on equal footing, we will fix the integrated **neutron fluence**:

$$
\tau \equiv \int n_n v_T\, dt \ \ (\text{units: mb}^{-1}=10^{-27}\,\mathrm{cm^2\ (per\ mb)}\,{}^{-1})
$$

and solve for the irradiation duration $t_{\rm irr}=\tau/(n_n v_T)$ at each chosen $n_n$. This isolates the effect of density itself (branch-point competition) from the trivial effect of "more neutrons capture more often." The literature value quoted for the main $s$-process component, accumulated over an AGB star's entire multi-thousand-pulse history, is $\tau_0\approx0.3\ \mathrm{mb}^{-1}$ (Käppeler et al. 2011) — but that number is large enough to drive every species in this small, large-cross-section Cs–Ba network to essentially complete depletion (try it in Exercise 1's code cell if you like: set `tau_target_mb_inv=0.3` and watch everything except $^{138}\mathrm{Ba}$ round to zero). That total is built up over many separate, much smaller-exposure pulses with dilution/mixing in between, which we are not modelling here. For a single-episode, single-zone toy network like this one, we instead use a smaller reference fluence, $\tau_0 = 0.01\ \mathrm{mb^{-1}}$, comparable to a single thermal-pulse-scale irradiation — large enough to move material well past $^{133}\mathrm{Cs}$, small enough that the whole isotope pattern (not just $^{138}\mathrm{Ba}$) remains visible. 


```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Thermal velocity at kT = 30 keV (standard s-process reference energy)
v_T = 2.194e8   # cm/s  (Bao et al. 2000)

MB = 1e-27      # 1 millibarn in cm^2
YR = 3.1557e7   # seconds per year
DAY = 86400.0   # seconds per day
```

## Building the network

**Species**:

| idx | nuclide | role |
|---|---|---|
| 0 | $^{133}\mathrm{Cs}$ | stable seed |
| 1 | $^{134}\mathrm{Cs}$ | branch point #1 ($t_{1/2}=2.0648$ yr) |
| 2 | $^{135}\mathrm{Cs}$ | quasi-stable on human timescales ($t_{1/2}=2.3\times10^6$ yr) |
| 3 | $^{136}\mathrm{Cs}$ | branch point #2 ($t_{1/2}=13.16$ d) |
| 4 | $^{137}\mathrm{Cs}$ | feeds $^{137}$Ba ($t_{1/2}=30.07$ yr) |
| 5 | $^{134}\mathrm{Ba}$ | stable, *s*-only |
| 6 | $^{135}\mathrm{Ba}$ | stable, mixed s+r |
| 7 | $^{136}\mathrm{Ba}$ | stable, *s*-only |
| 8 | $^{137}\mathrm{Ba}$ | stable, mixed s+r |
| 9 | $^{138}\mathrm{Ba}$ | stable, mixed s+r, $N=82$ **bottleneck** |
| 10 | "leak" ($^{139}\mathrm{La}$+) | sink beyond $^{138}\mathrm{Ba}$, tracks pile-up |

```python
species = ["Cs133","Cs134","Cs135","Cs136","Cs137", "Ba134","Ba135","Ba136","Ba137","Ba138","leak"]
idx = {name: i for i, name in enumerate(species)}
N = len(species)

# maxwellian-averaged cross sections at kT = 30 keV in mb 
sigma_mb = {"Cs133": 502.0, "Cs134": 142.0, "Cs135": 119.0, "Cs136": 155.0, "Cs137": 82.0, "Ba134": 425.0, "Ba135": 445.0, "Ba136": 62.0, "Ba137": 47.0, "Ba138": 4.0}

# beta-decay half-lives in seconds
thalf_s = {"Cs134": 2.0648*YR, "Cs135": 2.3e6*YR, "Cs136": 13.16*DAY, "Cs137": 30.07*YR}
lam_beta = {k: np.log(2)/t for k, t in thalf_s.items()}

print("Beta-decay rates (1/s):")
for k, v in lam_beta.items():
    print(f"  lambda_beta({k:6s}) = {v:.3e} 1/s   (t1/2 = {thalf_s[k]/YR:.3g} yr)")
```

## The ODE right-hand side

Each capture reaction contributes a loss term $-\lambda_n^{(i)} y_i$ to its parent and a gain term $+\lambda_n^{(i)} y_i$ to its daughter; each $\beta$-decay contributes $-\lambda_\beta y_i$ to the parent and $+\lambda_\beta y_i$ to the daughter. `build_rhs(n_n)` below builds this system for a given neutron density $n_n$ (cm$^{-3}$) and returns a function usable directly by `scipy.integrate.solve_ivp`.

```python
def capture_rate(nuclide, n_n):
    '''lambda_n = n_n * sigma_30keV * v_T, sigma converted mb -> cm^2'''
    return n_n * sigma_mb[nuclide] * MB * v_T

def build_rhs(n_n):
    ln = {s: capture_rate(s, n_n) for s in sigma_mb}
    def rhs(t, y):
        dydt = np.zeros(N)

        def cap(parent, daughter):
            r = ln[parent] * y[idx[parent]]
            dydt[idx[parent]]   -= r
            dydt[idx[daughter]] += r

        def dec(parent, daughter):
            r = lam_beta[parent] * y[idx[parent]]
            dydt[idx[parent]]   -= r
            dydt[idx[daughter]] += r

        # Cs chain with two branch points
        cap("Cs133", "Cs134")
        dec("Cs134", "Ba134");  cap("Cs134", "Cs135")     # branch point 1
        dec("Cs135", "Ba135");  cap("Cs135", "Cs136")     # negligible-decay "branch" (see Sec. 5)
        dec("Cs136", "Ba136");  cap("Cs136", "Cs137")     # branch point 2
        dec("Cs137", "Ba137")                             # Cs137 capture cross section not tracked (negligible path)

        # Ba isotopes keep capturing neutrons up the chain
        cap("Ba134", "Ba135")
        cap("Ba135", "Ba136")
        cap("Ba136", "Ba137")
        cap("Ba137", "Ba138")
        cap("Ba138", "leak")   # bottleneck: pile-up here because sigma is tiny

        return dydt
    return rhs

def run_network(n_n, tau_target_mb_inv=0.01, n_eval=400):
    '''Integrate the network at fixed neutron density n_n (cm^-3) up to the
    irradiation time needed to reach fluence tau_target (mb^-1).'''
    t_irr = tau_target_mb_inv / (n_n * v_T * MB)     # seconds
    y0 = np.zeros(N); y0[idx["Cs133"]] = 1.0
    t_eval = np.linspace(0, t_irr, n_eval)
    sol = solve_ivp(build_rhs(n_n), [0, t_irr], y0, t_eval=t_eval,
                     method="Radau", rtol=1e-9, atol=1e-14)
    return sol, t_irr
```

## First run: the $^{13}\mathrm{C}$-pocket

Run the network at $n_n=10^{8}\,\mathrm{cm^{-3}}$, a canonical value for the $^{13}\mathrm{C}$-pocket *s*-process neutron source in AGB stars at our reference single-episode exposure $\tau_0=0.01\ \mathrm{mb^{-1}}$.

```python
sol, t_irr = run_network(n_n=1e8)
print(f"Irradiation time at n_n = 1e8 cm^-3: {t_irr/YR:.2f} yr")

fig, ax = plt.subplots(1, 2, figsize=(12,4.5))

for s in ["Cs133","Cs134","Cs135","Cs136","Cs137"]:
    ax[0].plot(sol.t/YR, sol.y[idx[s]], label=s)
ax[0].set_xlabel("time (yr)"); ax[0].set_ylabel("abundance (rel.)")
ax[0].set_title("Cs isotopes vs time"); ax[0].legend(); ax[0].set_yscale("log")
ax[0].set_ylim(1e-8, 2)

final = sol.y[:, -1]
ba_names = ["Ba134","Ba135","Ba136","Ba137","Ba138"]
ba_vals = [final[idx[s]] for s in ba_names]
ax[1].bar(ba_names, ba_vals, color="steelblue")
ax[1].set_ylabel("final abundance (rel.)")
ax[1].set_title(f"Ba pattern at n$_n$=1e8 cm$^{{-3}}$ (classical s-process)")
plt.tight_layout(); plt.show()

print("\nFinal Ba pattern (normalized to Ba138):")
for s in ba_names:
    print(f"  {s}: {final[idx[s]]/final[idx['Ba138']]:.4f}")
```

**Check:** at this modest density and single-episode exposure, $^{136}\mathrm{Ba}$ should come out as the *largest* carrier, with $^{138}\mathrm{Ba}$ comparatively small — because $^{134}\mathrm{Cs}$ mostly decays (feeding $^{134}\mathrm{Ba}$, only lightly processed onward) while $^{136}\mathrm{Cs}$ decays even more completely (its 13.16-day half-life easily outruns capture at $n_n=10^8\,\mathrm{cm^{-3}}$, so essentially all of it becomes $^{136}\mathrm{Ba}$ rather than pushing on to $^{137}\mathrm{Cs}$). Very little material has yet fought its way past the $N=82$ bottleneck onto $^{138}\mathrm{Ba}$. ## Exercise 2 — scan neutron density and watch the bypass happen in stages

Now run the full network across the s-process → i-process density range at fixed fluence $\tau_0=0.01\ \mathrm{mb^{-1}}$, and track how the final Ba isotope pattern changes. Based on Exercise 1, you should expect the bypass to happen in **two separate stages**, because the two branch points turn on at very different densities:

- $^{134}\mathrm{Cs}$ ($t_{1/2}=2.06$ yr) already starts shifting toward capture — i.e. bypassing direct $^{134}$Ba production — at ordinary, even sub-*i*-process densities.
- $^{136}\mathrm{Cs}$ ($t_{1/2}=13.16$ d) has a much shorter half-life, so it takes a much higher density before capture can compete with its decay — i.e. $^{136}\mathrm{Ba}$ only starts getting bypassed, and $^{137}\mathrm{Ba}$ (via $^{137}\mathrm{Cs}$) only starts getting made, deep in the *i*-process regime.

This staged-bypass picture is exactly what is reported in the observational/presolar-grain literature: moderately enhanced neutron fluxes are known to "partially bypass both $^{134}\mathrm{Ba}$ and $^{135}\mathrm{Ba}$, and accumulate $^{136}\mathrm{Ba}$" (Liu et al. 2014, *ApJ*, 786, 66, on presolar SiC grain data), while the further step — an additional, density-driven rise in $^{137}\mathrm{Ba}/^{138}\mathrm{Ba}$ on top of the baseline pile-up already set by the $N{=}82$ bottleneck — is specifically identified as a true *i*-process signature, "at variance with the s process" (Reifarth et al., n\_TOF review). Let's see if the toy network reproduces this staging.


```python
n_n_scan = np.logspace(7, 17, 40)
ba_ratios = {s: [] for s in ba_names}

for n_n in n_n_scan:
    sol, _ = run_network(n_n)
    f = sol.y[:, -1]
    for s in ba_names:
        ba_ratios[s].append(f[idx[s]] / f[idx["Ba138"]])

fig, ax = plt.subplots(figsize=(7.5,5.5))
ax.plot(n_n_scan, ba_ratios["Ba135"], "o-", ms=3, label=r"$^{135}$Ba / $^{138}$Ba")
ax.plot(n_n_scan, ba_ratios["Ba137"], "s-", ms=3, label=r"$^{137}$Ba / $^{138}$Ba")
ax.plot(n_n_scan, ba_ratios["Ba136"], "^-", ms=3, label=r"$^{136}$Ba / $^{138}$Ba")
ax.plot(n_n_scan, ba_ratios["Ba134"], "d-", ms=3, label=r"$^{134}$Ba / $^{138}$Ba")
ax.set_xscale("log"); ax.set_yscale("log")

# mark the four physical stellar mechanisms from Section 1 directly on the plot
for label, n_n, color in [("13C pocket", 1e7, "tab:green"),
                           ("22Ne burst\n(low)", 3e8, "tab:green"),
                           ("22Ne burst\n(high)", 1e11, "tab:green"),
                           ("PIE / i-process", 4.3e14, "tab:red")]:
    ax.axvline(n_n, color=color, ls="--", alpha=0.6, lw=1)
ax.set_xlabel(r"neutron density $n_n$ (cm$^{-3}$)")
ax.set_ylabel("isotope ratio (relative to $^{138}$Ba)")
ax.set_title("Ba isotope pattern vs. neutron density\n(dashed lines: the four stellar mechanisms from Section 1)")
ax.legend(loc="upper left", fontsize=9)
plt.tight_layout(); plt.show()
```
